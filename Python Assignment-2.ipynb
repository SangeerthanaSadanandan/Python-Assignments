## <b> <span style="color:purple">List (Creation, Modification and Access):</span> </b>

### <b>1. List Creation: </b>

<b>a. Create a list named age_list with five integer elements. For eg., [24, 25, 26, 27, 28] </b>

```python
age_list=[24,25,26,27,28]
print(age_list)
```

##### <span style="color:green">The age_list is created to store five integer values representing the ages of individuals.The list allows multiple age values to be stored and accessed using a single variable.</span>

<b>b. Create a list named name_list with five string elements.  </b>

```python
name_list=["Michele","Peter","Rachel","Ross","Monica"]
print(name_list)
```

##### <span style="color:green">The name_list is created to store five string values. Each element represents the name of an individual. The list provides an organized way to store and access multiple names using a single variable.</span>

### <b>2. List Operations / Modifications: </b>

<b>a. Append the string "Yazhini" to name_list. </b>

```python
name_list.append("Yazhini")
print(name_list)
```

##### <span style="color:green">The append() method is used to add "Yazhini" as a new element at the end of the name_list.</span>

<b>b. Insert the element 30 at index 2 in age_list.  </b>

```python
age_list.insert(2,30)
print(age_list)
```

##### <span style="color:green">The insert() method is used to add the value 30 at index 2 of the age_list.</span>

<b>c. Remove the string "Yazhini" from name_list.</b>

```python
name_list.remove("Yazhini")
print(name_list)
```

##### <span style="color:green">The remove() method is used to delete the specified value "Yazhini" from the name_list.</span>

<b>d. Pop the last element from age_list. </b>

```python
age_list.pop()
print(age_list)
```

##### <span style="color:green">The pop() method is used to remove the last element from the age_list when no index is specified.</span>

<b>e. Extend the age_list with additional ages [29, 30, 26]. </b>

```python
age_list.extend([29,30,26])
print(age_list)
```

##### <span style="color:green">The extend() method is used to add multiple new elements [29, 30, 26] to the existing age_list.</span>

<b>f. Sort age_list in descending order.</b>

```python
age_list.sort(reverse=True)
print(age_list)
```

##### <span style="color:green">The sort() method with reverse=True is used to arrange the elements of age_list from highest to lowest.</span>

<b>g. Find Max age, Min age and sum of all ages from age_list. </b>

```python
Max_age=max(age_list)
Min_age=min(age_list)
Total_age=sum(age_list)
print("Maximum age:",Max_age)
print("Minimum age:",Min_age)
print("Sum of ages:",Total_age)
```

##### <span style="color:green">The max(), min(), and sum() functions are used to find the highest age, lowest age, and total of all the ages in the list.</span>

### <b>3. Accessing List Elements: </b>

<b>a. Print the first element of name_list.</b>

```python
print(name_list[0])
```

##### <span style="color:green">The first element is accessed using index 0, as Python list indexing starts from zero.</span>

<b>b. Print the last element of name_list. </b>

```python
print(name_list[-1])
```

##### <span style="color:green">The last element can be accessed using the index -1.</span>

<b>c. Print the elements from index 2 to index 4 in name_list.</b>

```python
print(name_list[2:5])
```

##### <span style="color:green">List slicing is used to access elements from index 2 through index 4. Since the ending index is excluded, 5 is used as the stopping index.</span>

<b>d. Print the elements of name_list in reverse order.</b>

```python
print(name_list[::-1])
```

##### <span style="color:green">The slicing notation [::-1] is used to reverse the order of elements in the list.</span>

## <b> <span style="color:purple">Dictionary (Creation, Modification and Access):</span> </b>

<b>a. Create a dictionary named student_marks that maps the names of five students to their marks (use scale of from 0 to 100).<b/>

```python
student_marks={
    "Michele":75,
    "Peter":88,
    "Rachel":92,
    "Ross":85,
    "Monica":78}
print(student_marks)
```

##### <span style="color:green">The student_marks dictionary is created to store the names of five students along with their marks. Each student's name is used as a key, and the corresponding mark is stored as its value.</span>

<b>b. Access and print the mark of a specific student, of your choice.</b>

```python
print("Mark of Rachel:",student_marks.get("Rachel"))
```

##### <span style="color:green">The get() method is used to access and display the mark of a specific student.</span>

<b>c. Add a new student "Janani" with a mark of 80 to the student_marks dictionary. </b>

```python
student_marks["Janani"]=80
print(student_marks)
```

##### <span style="color:green">A new key-value pair is added to the dictionary to store Janani's mark as 80.</span>

<b>d. Update the mark of any one older student to 82. </b>

```python
student_marks["Michele"]=82
print(student_marks)
```

##### <span style="color:green">The mark of an existing student is updated by assigning a new value to that student's key.</span>

<b>e. Use the keys(), values(), and items() methods to print all keys, values, and key-value pairs in the student_marks dictionary.</b> 

```python
print("All students:",student_marks.keys())
print("All marks:",student_marks.values())
print("All student-mark pairs:",student_marks.items())
```

##### <span style="color:green">The keys(), values(), and items() methods are used to access the keys, values, and complete key-value pairs in the dictionary.</span>

## <b> <span style="color:purple">Sets (Operations): </span> </b>

<b>a. Create a set called my_set with following values: ['a','e','i','o','u','a','a','i']. Analyse the output and provide explanation for the same.</b>

```python
my_set={"a","e","i","o","u","a","a","i"}
print(my_set)
```

##### Explanation:

##### <span style="color:green">A set does not allow duplicate values. Therefore, the repeated 'a' and 'i' are automatically removed.</span>

<b>b. Attempt to change the value of my_set[4] = 's'. If code throws an error, provide an explanation.<b/> 

```python
my_set[4]="s"
print(my_set)
```

##### <span style="color:green">Here i got an error,because sets are unordered and do not support indexing.Therefore, we cannot access or change an element using specific index or position.</span>

<b>c. Create two sets:  
set1 with values: {1, 3, 5, 7, 9}  
set2 with values: {2, 3, 5, 8, 10}</b>

```python
set1={1,3,5,7,9}
set2={2,3,5,8,10}
print(set1)
print(set2)
```

##### <span style="color:green">Two sets are created with the given integer values.</span>

<b>d. Compute and print the union and intersection of set1 and set2.</b>

```python
union_set=set1.union(set2)
intersection_set = set1.intersection(set2)
print("Union:", union_set)
print("Intersection:", intersection_set)
```

##### <span style="color:green">The union() method combines all unique elements from both sets. The intersection() method returns only the elements that are common to both sets.</span>

### <b>Performance Category Program:  </b>

<b>1. Prompt user for Input. Score range should be from 0 to 10 (both inclusive).</b>

<b>2. Find the performance category based on the input score using following criteria:</b>
   
<b>a. Above Average:  Score greater than 7 </b>

<b>b. Average: Score between 4 and 7(both inclusive) </b>

<b>c. Below Average: Score lesser than 4 </b>

<b>4. Output: Print the Performance category</b>
   
<b>6. Additional Step: You can give a prompt of your choice to each category.</b> 
For eg: If score below average “Need to Improve your performance, consistent practice will lead to better results”.  

```python
print("Performance Category Program: ")
score=int(input("Enter your score (0 to 10):"))
if score>7:
    print("Above Average: Excellent work!")
elif score >= 4:
    print("Good effort! Keep practicing, there's room for improvement.")
else:
    print("Need to Improve your performance, consistent practice will lead to better results.")
```

##### <span style="color:green">The program accepts a score from the user and uses if, elif, and else statements to determine the performance category based on the given score.</span>
