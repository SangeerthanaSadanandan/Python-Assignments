### <b>Number Guessing Game</b>

Problem Statement: Create a Python program that implements a simple number 
guessing game using a while loop. The program should make use of control 
statements such as else, break, and continue. 

<b>1. Set Up the Game:</b>

   ★ Generate a random number between 1 and 10 that the user has to guess. 
      Import random and use randint function.
      
<b>2. Prompt the User:</b>

   ★ Ask the user to guess the number.
   
   ★ Set a variable attempts to 3, which represents the maximum number of 
      guesses allowed.
      
<b>3. Implement the Guessing Logic:</b>

   ★ Use a while loop to allow the user to keep guessing until they get the 
     correct number or run out of attempts.
     
   ★ Provide feedback to the user for each guess:
   
       ➢ If the guess is out of the valid range (1 to 10), inform the user. 
       ➢ If the guess is greater than the secret number. 
       ➢ If the guess is lower than the secret number. 
       ➢ If the guess is correct, congratulate the user and end the game.
       
<b>4. Control Statements:</b>

   ★ Use continue to skip the rest of the loop after informing the user if the 
     guess is out of range.
     
   ★ Use break to exit the loop when the guess is correct.
   
   ★ Use else with the while loop to provide a message like "Better luck next 
      time!" if the user runs out of attempts without guessing the correct number. 


```python
import random
secret_number=random.randint(1,10)
attempts=3
while attempts>0:
    guess=int(input("Guess the number (between 1 and 10): "))
    if guess<1 or guess>10:
              print("Your guess is out of range. Please guess a number between 1 and 10.")
              continue
    attempts-=1
    if guess==secret_number:
        print("Congratulations! You guessed the correct number.")
        break
    elif guess>secret_number:
        print("Too high. Try again.")
    else:
        print("Too low. Try again.")
else:
    print("Your chances are over...Better luck next time!\nThe correct number was: ",secret_number)
        
```

##### <span style="color:green">This is a simple Number Guessing Game. The random.randint() function generates a random number between 1 and 10.The user gets three attempts to guess the correct number.</span>

##### <span style="color:green">A while loop is used to repeatedly ask the user for a guess. It checks whether the entered number is within the range of 1 to 10. If the number is outside this range, continue skips the remaining statements and asks for another guess.</span>

##### <span style="color:green">The if, elif, and else statements compare the user's guess with the secret number and display whether the guess is too high or too low. If the user guesses correctly, break immediately exits the loop.</span>

##### <span style="color:green">The else block of the while loop executes when all valid attempts are used without finding the correct number. It displays a "Better luck next time!" message and correct number.</span>

### <b>Multiplication Table Generator </b>

Problem Statement: Create a Python program that generates and prints a 
multiplication table (from 1 to 10) for a given number using a for loop and the range 
function.

1. Prompt user for Input. Ask the user to enter a number for which they want to 
generate a multiplication table. 

2. Generate the Multiplication Table: Use a for loop to iterate through the numbers 
1 to 10. In each iteration, calculate the product of the user's number and the 
current number from the loop. 

3. Display the Multiplication Table:  
Print each line of the multiplication table in the format: "number x i = result".

```python
number=int(input("Enter the number for which you want the multiplication table: "))
for i in range(1,11):
    result=number*i
    print(number,"x",i,"=",result)
    
```

##### <span style="color:green">This program generates a multiplication table from 1 to 10 for a number entered by the user.The input() function is used to get a number from the user, and int() converts the input into an integer and stores it in the variable "number". </span>
##### <span style="color:green">A for loop is used to repeat the multiplication process. The range(1, 11) function generates numbers from 1 to 10. In each iteration, the user's number is multiplied by the current value of i and stored in the variable "result".</span>
##### <span style="color:green">Finally, the result is displayed in the format: number x i = result</span>

### <b>BMI Calculator</b>

Problem Statement: Create a Python program that calculates the Body Mass Index 
(BMI).

1. Define a function calculate_bmi(weight, height) that returns the BMI.

2. Prompt the user for their weight (in kg) and height (in meters).

3. Use the function to calculate and display the BMI 

```python
def calculate_bmi(weight,height):
    BMI=round(weight/(height**2),2)
    return BMI

weight=float(input("Enter your weight in kg: "))
height=float(input("Enter your height in meters: "))

BMI=calculate_bmi(weight,height)
print("Your BMI is:", BMI)
```

##### <span style="color:green">The function calculate_bmi(weight, height) takes the user's weight in kilograms and height in meters as parameters. It calculates BMI using the formula: </span></b>
##### <span style="color:green"> BMI = weight/[height]²  </span></b>
##### <span style="color:green"> The return statement sends the calculated BMI back to the main program. The program then displays the BMI rounded to two decimal places.</span></b>
